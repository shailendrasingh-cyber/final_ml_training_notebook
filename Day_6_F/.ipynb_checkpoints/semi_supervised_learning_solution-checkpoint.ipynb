# Semi-Supervised Learning: From Missing Labels to a Predictive Model

**Beginner-friendly practical lecture | 100,000 records**

## What will we learn?
1. What supervised, unsupervised and semi-supervised learning mean.
2. How to inspect a dataset containing both known and missing target labels.
3. How to scale numerical features.
4. How to use the **Elbow Method** and **Silhouette Score** to study the number of clusters.
5. How to use K-Means clusters to infer missing target labels (pseudo-labeling).
6. How to train a Machine Learning classifier using real labels plus inferred labels.
7. How to calculate accuracy on labels that were kept aside and never used for training.
8. How to use PCA to reduce eight features to two dimensions and understand explained variance.

> Important: the main dataset has exactly 100,000 rows. Only 30% of its target values are visible; the remaining 70% are blank. The separate answer-key file is used only for final educational evaluation. Never use it to train or infer labels.

## 1. Understand the idea

- **Supervised learning:** the computer learns from examples where the input and correct answer are both available.
- **Unsupervised learning:** the computer receives inputs without answers and tries to find structure, such as groups.
- **Semi-supervised learning:** we combine a small amount of labelled data with a much larger amount of unlabelled data.

### Example
Imagine 100,000 customer records. We know the customer category for only 30,000 records. The category is missing for 70,000 records. We use the known examples and patterns in the data to estimate the missing categories.

In this practical, the missing answers are called **pseudo-labels**. They are predictions, not guaranteed correct answers.

## 2. Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA

RANDOM_STATE = 42
```

## 3. Load the dataset

```python
# Keep the CSV file in the same folder as this notebook.
df = pd.read_csv("semi_supervised_dataset_100k.csv")

print("Dataset shape:", df.shape)
display(df.head())
print("\\nMissing target values:", df["target"].isna().sum())
print("Known target values:", df["target"].notna().sum())
print("\\nTarget distribution among known labels:")
display(df["target"].value_counts(dropna=False).sort_index())
```

### What do these numbers mean?
- `100000` rows means there are 100,000 records.
- `record_id` is only an identifier; it is not a learning feature.
- `feature_1` to `feature_8` are the input variables.
- `target` is the answer we want the model to learn.
- A blank/NaN target means the answer is not supplied.

We must remove `record_id` from model input because it is an identifier, not a meaningful measurement.

## 4. Separate labelled and unlabelled records

```python
feature_cols = [f"feature_{i}" for i in range(1, 9)]

labelled = df[df["target"].notna()].copy()
unlabelled = df[df["target"].isna()].copy()

X_labelled = labelled[feature_cols]
y_labelled = labelled["target"].astype(int)
X_unlabelled = unlabelled[feature_cols]

print("Labelled records:", len(labelled))
print("Unlabelled records:", len(unlabelled))
print("Total:", len(labelled) + len(unlabelled))
```

## 5. Split known labels into training and validation sets

We reserve 20% of the known labels for testing. Their answers are hidden from the training process. This gives us a fair way to check how well the final model predicts unseen examples.

The unlabelled records are not part of this test set; they are the records whose target values we want to estimate.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_labelled, y_labelled,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y_labelled
)

print("Known labels used for training:", len(X_train))
print("Known labels kept for testing:", len(X_test))
```

## 6. Feature scaling

K-Means uses distance. A feature with a very large numerical range can dominate the distance calculation. StandardScaler transforms each feature so that it is approximately centred around 0 with a standard deviation of 1.

**Formula:**

`z = (x - mean) / standard_deviation`

Fit the scaler only on the training data, then use the same scaler to transform all other data. This prevents information from the test set influencing the scaling step.

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_unlabelled_scaled = scaler.transform(X_unlabelled)

# All records for exploratory clustering: training labels + unlabelled records.
X_pool_scaled = np.vstack([X_train_scaled, X_unlabelled_scaled])
print("Scaled pool shape:", X_pool_scaled.shape)
```

## 7. Find a suitable cluster count: Elbow Method

K-Means groups records into `K` clusters. The Elbow Method plots **inertia** against K.

- Inertia is the sum of squared distances from each record to its assigned cluster centre.
- Inertia always decreases as K increases.
- We look for the point where the reduction starts slowing down. This bend is called the elbow.

The elbow is a visual guide, not an absolute mathematical answer.

```python
k_values = range(2, 9)
inertias = []

# Use a reproducible sample to make the cluster-count experiment faster.
rng = np.random.default_rng(RANDOM_STATE)
sample_size = min(12000, len(X_pool_scaled))
sample_indices = rng.choice(len(X_pool_scaled), size=sample_size, replace=False)
X_cluster_sample = X_pool_scaled[sample_indices]

for k in k_values:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    km.fit(X_cluster_sample)
    inertias.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(list(k_values), inertias, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

### How to read the elbow graph
Look for a bend where the line changes from a steep fall to a slower fall. For example, if the curve falls sharply up to K=4 and then becomes flatter, K=4 may be a reasonable candidate.

Do not select K only because it gives the smallest inertia. A very large K will almost always reduce inertia.

## 8. Check cluster quality: Silhouette Score

The Silhouette Score measures how well each record fits its own cluster compared with nearby clusters.

- Close to **+1**: clusters are well separated.
- Around **0**: clusters overlap.
- Below **0**: some records may be assigned to the wrong cluster.

We calculate it on a sample because silhouette calculations can be expensive on very large datasets.

```python
silhouette_scores = []

for k in k_values:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    cluster_labels = km.fit_predict(X_cluster_sample)
    score = silhouette_score(X_cluster_sample, cluster_labels)
    silhouette_scores.append(score)
    print(f"K={k}: silhouette score = {score:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(list(k_values), silhouette_scores, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()

best_k_silhouette = list(k_values)[int(np.argmax(silhouette_scores))]
print("K with highest silhouette score:", best_k_silhouette)
```

### Decide K using both methods
Compare the elbow bend and the silhouette-score graph. The highest silhouette score is one useful signal, while the elbow gives a second, visual signal. If they suggest different values, discuss the trade-off and choose a sensible K based on both plots.

For this teaching dataset, four natural groups were generated. The three target categories are not exactly the same thing as the four clusters: two natural groups share one target category.

## 9. Fit K-Means and infer the missing target labels

We use K-Means to assign a cluster ID to each labelled training record and each unlabelled record. Then, for every cluster, we find the most common **known target class** in that cluster.

That majority class becomes the cluster's pseudo-label. We only assign pseudo-labels to unlabelled records in clusters where the most common known class is sufficiently dominant.

**Confidence threshold:** 0.70 means at least 70% of the known labels in that cluster must agree on the same class. This is a simple teaching approach; real projects should compare more advanced semi-supervised methods too.

```python
# Use the selected K from the combined elbow + silhouette discussion.
# Change this value after inspecting both graphs if needed.
chosen_k = best_k_silhouette
print("Chosen K:", chosen_k)

kmeans = KMeans(n_clusters=chosen_k, random_state=RANDOM_STATE, n_init=10)
train_clusters = kmeans.fit_predict(X_train_scaled)
unlabelled_clusters = kmeans.predict(X_unlabelled_scaled)

cluster_map = {}
cluster_confidence = {}

for cluster_id in range(chosen_k):
    labels_in_cluster = y_train.to_numpy()[train_clusters == cluster_id]
    if len(labels_in_cluster) == 0:
        continue
    counts = pd.Series(labels_in_cluster).value_counts(normalize=True)
    cluster_map[cluster_id] = int(counts.index[0])
    cluster_confidence[cluster_id] = float(counts.iloc[0])

print("Cluster -> pseudo-label mapping:", cluster_map)
print("Cluster confidence:", cluster_confidence)
```

```python
CONFIDENCE_THRESHOLD = 0.70

pseudo_labels = []
accepted_positions = []

for position, cluster_id in enumerate(unlabelled_clusters):
    confidence = cluster_confidence.get(int(cluster_id), 0)
    if confidence >= CONFIDENCE_THRESHOLD and int(cluster_id) in cluster_map:
        accepted_positions.append(position)
        pseudo_labels.append(cluster_map[int(cluster_id)])

X_pseudo = X_unlabelled.iloc[accepted_positions].copy()
y_pseudo = np.array(pseudo_labels, dtype=int)

print("Unlabelled records:", len(X_unlabelled))
print("Records receiving pseudo-labels:", len(X_pseudo))
print("Not pseudo-labelled due to low confidence:", len(X_unlabelled) - len(X_pseudo))
print("Pseudo-label distribution:")
print(pd.Series(y_pseudo).value_counts().sort_index())
```

### Important limitation
Every record in one cluster receives the same majority-class pseudo-label in this basic method. If a cluster contains mixed target classes, some pseudo-labels can be wrong. The confidence threshold reduces uncertain assignments but does not guarantee correctness.

This is why we evaluate the final classifier on the separate known-label test set.

## 10. Train a Machine Learning model using real + pseudo labels

Now we train a Random Forest classifier using:
- Real labelled training examples (`X_train`, `y_train`)
- Unlabelled examples that received pseudo-labels (`X_pseudo`, `y_pseudo`)

The validation/test examples remain separate and are not used to fit the model.

```python
X_combined = pd.concat([X_train, X_pseudo], axis=0)
y_combined = np.concatenate([y_train.to_numpy(), y_pseudo])

final_model = RandomForestClassifier(
    n_estimators=150,
    max_depth=None,
    min_samples_split=2,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    class_weight="balanced"
)
final_model.fit(X_combined, y_combined)

y_pred = final_model.predict(X_test)
final_accuracy = accuracy_score(y_test, y_pred)

print(f"Final test accuracy: {final_accuracy:.4f}")
print(f"Final test accuracy (%): {final_accuracy * 100:.2f}%")
print("\\nClassification report:")
print(classification_report(y_test, y_pred))
```

## 11. Compare against a model trained only on real labels

This comparison helps us understand whether adding pseudo-labelled records improved the result. Both models use the exact same held-out test set.

```python
baseline_model = RandomForestClassifier(
    n_estimators=150,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    class_weight="balanced"
)
baseline_model.fit(X_train, y_train)
baseline_pred = baseline_model.predict(X_test)
baseline_accuracy = accuracy_score(y_test, baseline_pred)

comparison = pd.DataFrame({
    "Model": ["Only real labelled data", "Real + pseudo-labelled data"],
    "Test accuracy": [baseline_accuracy, final_accuracy],
    "Accuracy (%)": [baseline_accuracy * 100, final_accuracy * 100]
})
display(comparison)
```

### Understand accuracy
`Accuracy = Correct predictions / Total test predictions`

For example, if a model correctly predicts 900 out of 1,000 test records:

`Accuracy = 900 / 1000 = 0.90 = 90%`

Accuracy is not the only metric. If classes are imbalanced, also inspect precision, recall and F1-score in the classification report.

## 12. PCA analysis: reduce 8 features to 2

**Principal Component Analysis (PCA)** is a dimensionality-reduction technique. It combines the original features into new variables called principal components.

- PC1 captures the largest possible amount of variation.
- PC2 captures the next largest amount, while being perpendicular to PC1.
- We use two components so that the data can be plotted on a 2D graph.

PCA is fitted on the training data only, then applied to the test data. The graph below colours the test records by their actual known class, only for visual interpretation.

```python
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance explained by 2 components:",
      pca.explained_variance_ratio_.sum() * 100, "%")

plt.figure(figsize=(8, 6))
scatter = plt.scatter(
    X_test_pca[:, 0], X_test_pca[:, 1],
    c=y_test, cmap="viridis", s=10, alpha=0.6
)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Test Records in Two Dimensions")
plt.colorbar(scatter, label="Known target class")
plt.grid(True)
plt.show()
```

### PCA interpretation
If the first two components explain 75% of the variance, it means the 2D representation preserves 75% of the variation measured across the original eight features. It does **not** mean the model is 75% accurate.

PCA is mainly used here for visual understanding. The Random Forest model above was trained on the original eight features.

## 13. Confusion matrix

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
plt.imshow(cm)
plt.title("Confusion Matrix - Final Model")
plt.xlabel("Predicted class")
plt.ylabel("Actual class")
plt.xticks(range(len(np.unique(y_test))))
plt.yticks(range(len(np.unique(y_test))))
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")
plt.colorbar()
plt.show()
```

## 14. Save inferred target values

The following creates a new CSV containing the original records and the inferred labels. The original `target` column is preserved, and a separate `predicted_target` column is added.

Records below the confidence threshold remain blank in `predicted_target`.

```python
result = df.copy()
result["predicted_target"] = result["target"]

unlabelled_record_ids = unlabelled.iloc[accepted_positions]["record_id"].to_numpy()
prediction_lookup = dict(zip(unlabelled_record_ids, y_pseudo))
result.loc[result["record_id"].isin(prediction_lookup), "predicted_target"] = (
    result.loc[result["record_id"].isin(prediction_lookup), "record_id"].map(prediction_lookup)
)

result.to_csv("semi_supervised_results.csv", index=False)
print("Saved: semi_supervised_results.csv")
display(result.head())
```
