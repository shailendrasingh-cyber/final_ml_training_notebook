# Lecture 1: Introduction to Semi-Supervised Learning
### From Missing Target Labels (`NaN`) to a Machine Learning Classifier

**Dataset:** Real-World Wine Chemical Analysis Dataset (Secondary Dataset — 178 wine samples, 13 chemical measurements)

---

## What Will We Learn Today?
1. **Core Concept:** The difference between Supervised, Unsupervised, and Semi-Supervised Learning.
2. **Data Inspection:** Exploring a real-world dataset where **only 30%** of the target labels are known and **70% are missing (`NaN`)**.
3. **Data Preparation:** Separating known vs. unknown rows, creating a fair test split, and scaling features.
4. **Finding Clusters (Unsupervised Step):** Using the **Elbow Method** and **Silhouette Score** to find the best number of clusters ($K$).
5. **Pseudo-Labeling:** Filling in the missing `NaN` targets using K-Means cluster majority voting.
6. **Model Training (Supervised Step):** Training a Random Forest classifier on **Known Labels + Pseudo-Labels**.
7. **Evaluation & Visualization:** Checking accuracy on unseen test data and visualizing clusters with **PCA**.

## Part 1: Understanding Semi-Supervised Learning (Simple Analogy)

- **Supervised Learning (Teacher-Guided):** Every example has a clear answer label (e.g., 100% of wines are labeled as Type 0, Type 1, or Type 2).
- **Unsupervised Learning (Self-Discovery):** No answers exist at all. The algorithm groups similar items together based purely on their features.
- **Semi-Supervised Learning (Best of Both Worlds):** In real life, labeling data is expensive and slow. Usually, we have **a small amount of labeled data** and **a large amount of unlabeled (`NaN`) data**.

### Today's Classroom Story:
Imagine a wine laboratory tested **178 bottles** for 13 chemical properties (`alcohol`, `color_intensity`, `proline`, etc.). An expert sommelier only had time to label **53 bottles (30%)**. The remaining **125 bottles (70%)** have `NaN` (missing) in the `wine_cultivar` column.

**Our Goal:** Use the patterns in all 178 bottles to group similar wines together, fill in the missing `NaN` labels (**pseudo-labels**), and train a Machine Learning model!

## Part 2: Step-by-Step Python Implementation
*(Every step is placed in its own cell so we can run and discuss each line together.)*

### Step 1: Import the Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, accuracy_score, classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA

RANDOM_STATE = 42
print("All libraries imported successfully!")
```

### Step 2: Load the Secondary Dataset
Make sure `wine_semi_supervised_dataset.csv` is in the same folder as this notebook.

```python
df = pd.read_csv("wine_semi_supervised_dataset.csv")
print("Dataset Shape (Rows, Columns):", df.shape)
df.head(10)
```

### Step 3: Check Known vs. Missing (`NaN`) Target Labels
Let's see how many rows have a known `wine_cultivar` label and how many are blank (`NaN`).

```python
known_count = df["wine_cultivar"].notna().sum()
missing_count = df["wine_cultivar"].isna().sum()

print(f"Total wine samples : {len(df)}")
print(f"Known labels       : {known_count} ({known_count/len(df)*100:.1f}%)")
print(f"Missing (NaN)      : {missing_count} ({missing_count/len(df)*100:.1f}%)")
print("\nBreakdown of target values:")
print(df["wine_cultivar"].value_counts(dropna=False).sort_index())
```

### Step 4: Separate Feature Columns from the Target Column
Remember: `sample_id` is just a row number, so we do **not** give it to the model.

```python
feature_cols = [col for col in df.columns if col not in ["sample_id", "wine_cultivar"]]

print(f"Number of input features: {len(feature_cols)}")
print("Feature names:", feature_cols)
```

### Step 5: Split the Data into Labelled Rows and Unlabelled (`NaN`) Rows

```python
labelled_df = df[df["wine_cultivar"].notna()].copy()
unlabelled_df = df[df["wine_cultivar"].isna()].copy()

X_labelled = labelled_df[feature_cols]
y_labelled = labelled_df["wine_cultivar"].astype(int)

X_unlabelled = unlabelled_df[feature_cols]

print("Labelled rows shape   :", X_labelled.shape)
print("Unlabelled rows shape :", X_unlabelled.shape)
```

### Step 6: Hold Out a Test Set from the Known Labels
Before we do any clustering or pseudo-labeling, we lock away **25% of the known labels** as our final exam (`X_test`, `y_test`). The model will never see these during training.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_labelled,
    y_labelled,
    test_size=0.25,
    random_state=RANDOM_STATE,
    stratify=y_labelled
)

print("Known samples for training :", len(X_train))
print("Known samples for testing  :", len(X_test))
```

### Step 7: Scale the Features (`StandardScaler`)
Why do we scale? In our wine dataset, `proline` ranges from 278 to 1680, while `nonflavanoid_phenols` is around 0.13 to 0.66. Without scaling, K-Means would only pay attention to `proline`!

```python
scaler = StandardScaler()

# Fit only on training data to avoid data leakage
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_unlabelled_scaled = scaler.transform(X_unlabelled)

# Combine training + unlabelled features for clustering
X_cluster_pool = np.vstack([X_train_scaled, X_unlabelled_scaled])

print("Total samples available for clustering:", X_cluster_pool.shape[0])
```

---
## Part 3: Finding the Best Number of Clusters ($K$)

### Step 8: Run the Elbow Method (Calculate Inertia for $K = 2$ to $8$)
**Inertia** measures how tightly packed each cluster is (lower inertia = tighter clusters). We look for the "elbow" where adding more clusters stops giving big improvements.

```python
k_range = range(2, 9)
inertias = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    km.fit(X_cluster_pool)
    inertias.append(km.inertia_)
    print(f"K = {k} -> Inertia = {km.inertia_:.2f}")
```

### Step 9: Plot the Elbow Curve

```python
plt.figure(figsize=(8, 4))
plt.plot(list(k_range), inertias, marker="o", linewidth=2, color="steelblue")
plt.title("Elbow Method: Finding the Bend in Inertia")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia (Within-Cluster Sum of Squares)")
plt.xticks(list(k_range))
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

### Step 10: Calculate Silhouette Scores for $K = 2$ to $8$
The **Silhouette Score** ranges from **-1 to +1**:
- **Higher score (closer to +1)** means clusters are well-separated and distinct.
- **Around 0** means clusters are overlapping.

```python
silhouette_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    labels = km.fit_predict(X_cluster_pool)
    score = silhouette_score(X_cluster_pool, labels)
    silhouette_scores.append(score)
    print(f"K = {k} -> Silhouette Score = {score:.4f}")
```

### Step 11: Plot the Silhouette Scores and Pick Optimal $K$

```python
plt.figure(figsize=(8, 4))
plt.plot(list(k_range), silhouette_scores, marker="s", linewidth=2, color="darkorange")
plt.title("Silhouette Score for Each K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(list(k_range))
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()

optimal_k = list(k_range)[int(np.argmax(silhouette_scores))]
print(f"Best K suggested by Elbow + Silhouette Score: K = {optimal_k}")
```

---
## Part 4: Filling the `NaN` Targets (Pseudo-Labeling)

### Step 12: Fit K-Means with Optimal $K$ and Assign Cluster IDs
Now we group both our known training samples and our unlabeled (`NaN`) samples into $K$ clusters.

```python
kmeans = KMeans(n_clusters=optimal_k, random_state=RANDOM_STATE, n_init=10)
kmeans.fit(X_cluster_pool)

train_clusters = kmeans.predict(X_train_scaled)
unlabelled_clusters = kmeans.predict(X_unlabelled_scaled)

print("Cluster IDs assigned to training samples   :", np.unique(train_clusters, return_counts=True))
print("Cluster IDs assigned to unlabelled samples :", np.unique(unlabelled_clusters, return_counts=True))
```

### Step 13: Map Each Cluster to Its Majority Known Target Class
We look inside each cluster, check the known labels (`y_train`), and find which wine class appears most often (majority vote) along with how confident that vote is.

```python
cluster_to_target = {}
cluster_confidence = {}

for cid in range(optimal_k):
    known_in_cluster = y_train.to_numpy()[train_clusters == cid]
    if len(known_in_cluster) > 0:
        proportions = pd.Series(known_in_cluster).value_counts(normalize=True)
        majority_class = int(proportions.index[0])
        confidence = float(proportions.iloc[0])
        cluster_to_target[cid] = majority_class
        cluster_confidence[cid] = confidence
        print(f"Cluster {cid} -> Majority Target: {majority_class} (Confidence: {confidence*100:.1f}%)")
```

### Step 14: Fill the Missing (`NaN`) Targets Using Pseudo-Labels
We only accept pseudo-labels if the cluster confidence is at least **70%** (`0.70`).

```python
CONFIDENCE_THRESHOLD = 0.70

accepted_indices = []
pseudo_y_list = []

for idx, cid in enumerate(unlabelled_clusters):
    cid = int(cid)
    if cid in cluster_to_target and cluster_confidence[cid] >= CONFIDENCE_THRESHOLD:
        accepted_indices.append(idx)
        pseudo_y_list.append(cluster_to_target[cid])

X_pseudo = X_unlabelled.iloc[accepted_indices].copy()
y_pseudo = np.array(pseudo_y_list, dtype=int)

# Also show the filled dataset so students can see NaNs replaced!
df_filled = df.copy()
filled_row_indices = unlabelled_df.iloc[accepted_indices].index
df_filled.loc[filled_row_indices, "wine_cultivar"] = y_pseudo

print(f"Total NaN rows originally : {len(X_unlabelled)}")
print(f"Rows filled with pseudo-labels : {len(X_pseudo)}")
print(f"Remaining NaN rows : {df_filled['wine_cultivar'].isna().sum()}")
df_filled.head(10)
```

---
## Part 5: Training & Evaluating the Machine Learning Classifier

### Step 15: Combine Real Labeled Data + Pseudo-Labeled Data

```python
X_semi_train = pd.concat([X_train, X_pseudo], axis=0)
y_semi_train = np.concatenate([y_train.to_numpy(), y_pseudo])

print("Original small training set size       :", len(X_train))
print("New semi-supervised training set size  :", len(X_semi_train))
```

### Step 16: Train the Semi-Supervised Random Forest Model and Test It

```python
semi_model = RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE)
semi_model.fit(X_semi_train, y_semi_train)

y_pred_semi = semi_model.predict(X_test)
semi_acc = accuracy_score(y_test, y_pred_semi)

print(f"Semi-Supervised Model Test Accuracy: {semi_acc * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, y_pred_semi))
```

### Step 17: Compare Supervised-Only vs. Semi-Supervised Learning
Let's see how a model trained **only** on the small 30% labeled set compares against our semi-supervised model!

```python
sup_model = RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE)
sup_model.fit(X_train, y_train)

y_pred_sup = sup_model.predict(X_test)
sup_acc = accuracy_score(y_test, y_pred_sup)

results_df = pd.DataFrame({
    "Approach": ["Supervised Only (Small Labeled Data)", "Semi-Supervised (Labeled + Pseudo-Labeled)"],
    "Training Samples Used": [len(X_train), len(X_semi_train)],
    "Test Accuracy (%)": [round(sup_acc * 100, 2), round(semi_acc * 100, 2)]
})
results_df
```

---
## Part 6: Visualizing the Clusters in 2D using PCA

### Step 18: Reduce 13 Features to 2 Principal Components and Plot

```python
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_all_scaled = scaler.transform(df[feature_cols])
X_pca = pca.fit_transform(X_all_scaled)
all_clusters = kmeans.predict(X_all_scaled)

plt.figure(figsize=(8, 5))
for cid in range(optimal_k):
    mask = (all_clusters == cid)
    plt.scatter(
        X_pca[mask, 0],
        X_pca[mask, 1],
        label=f"Cluster {cid} -> Target {cluster_to_target.get(cid, '?')}",
        alpha=0.8,
        s=50
    )

plt.title(f"PCA 2D Visualization of K-Means Clusters (Explained Variance: {pca.explained_variance_ratio_.sum()*100:.1f}%)")
plt.xlabel("Principal Component 1 (PC1)")
plt.ylabel("Principal Component 2 (PC2)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
