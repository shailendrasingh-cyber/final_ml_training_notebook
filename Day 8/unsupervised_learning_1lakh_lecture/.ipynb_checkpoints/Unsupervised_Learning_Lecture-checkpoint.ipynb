## Step 1️ — What is Unsupervised Learning?

**Easy example:** Imagine a basket full of mixed fruits, and nobody told you their names.
You will still put *similar fruits together*: all round red ones in one pile, long yellow ones in another.
That is **unsupervised learning**.

| | Supervised Learning | Unsupervised Learning |
|---|---|---|
| Data has answer (label)? | ✅ Yes | ❌ No |
| Goal | Predict the answer | Find hidden groups / patterns |
| Example | Spam or Not Spam | Group customers by behaviour |

**Our data has NO label column**, so we will let the computer find groups by itself.

## Step 2️— Main types of Unsupervised Learning

1. **Clustering** → make groups of similar rows (**we do this today with K-Means**)
2. **Dimensionality Reduction** → reduce many columns to 2 or 3 so we can draw them (**we do this with PCA**)
3. **Anomaly Detection** → find strange/odd rows

## Step 3️— Import the libraries
These are our tools: `pandas` (tables), `matplotlib` (graphs), `sklearn` (machine learning).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

print("All libraries imported ")
```

## Step 4️ — Load the data
Keep the CSV file in the **same folder** as this notebook.

```python
df = pd.read_csv("real_dataset_1lakh.csv")
print("Rows and Columns:", df.shape)
```

### Look at the first 5 rows

```python
df.head()
```

### What do the columns mean?

This is *forest land* data. Each row = one small piece of land.

| Column | Meaning |
|---|---|
| `Elevation` | Height above sea level (metres) |
| `Aspect` | Direction the slope faces (0–360°) |
| `Slope` | How steep the land is |
| `Horizontal_Distance_To_Hydrology` | Distance to nearest water |
| `Vertical_Distance_To_Hydrology` | Height difference from nearest water |
| `Horizontal_Distance_To_Roadways` | Distance to nearest road |
| `Hillshade_9am / Noon / 3pm` | How much sunlight/shade at that time (0–255) |
| `Horizontal_Distance_To_Fire_Points` | Distance to nearest wildfire point |
| `Wilderness_Area_0..3` | Which of 4 wilderness areas (0 or 1) |
| `Soil_Type_0..39` | Which of 40 soil types (0 or 1) |

### Check for missing values

```python
print("Total missing values:", df.isnull().sum().sum())
```

 Zero missing values, so no cleaning is needed. Now a quick summary of the numbers:

```python
df.iloc[:, :10].describe().round(1)
```

## Step 5️— Choose the columns (features) for clustering

To keep things **simple**, we use only the **first 10 columns** (the real measurements).
The remaining columns (Wilderness & Soil) are 0/1 flags, so we leave them for now.

```python
feature_cols = df.columns[:10].tolist()
X = df[feature_cols]
print("Columns used:")
for c in feature_cols:
    print(" -", c)
```

## Step 6️ — Scale the data (very important!)

Look at the problem: `Elevation` is around **3000**, but `Slope` is around **14**.
K-Means works with *distance*, so big numbers would dominate small numbers. Unfair!

**Solution:** `StandardScaler` converts every column to the same scale (mean = 0, spread = 1).

```python
print("BEFORE scaling (average of each column):")
print(X.mean().round(1))
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("AFTER scaling (average of each column):")
print(pd.Series(X_scaled.mean(axis=0), index=feature_cols).round(2))
```

## Step 7️ — How does K-Means work?

Think of it like this:
1. Choose **K** (how many groups you want)
2. Put K **centre points** randomly
3. Every row joins the **nearest centre**
4. Move each centre to the **middle of its group**
5. Repeat 3 and 4 until nothing changes 

**Big question: what should K be?** Let us find out with two simple tools.

### Tool 1 — Elbow Method
We try K = 1 to 8 and see how the error (*inertia*) falls. We look for the **bend (elbow)** in the line.

```python
inertia = []
k_range = range(1, 9)

for k in k_range:
    model = KMeans(n_clusters=k, n_init=10, random_state=42)
    model.fit(X_scaled)
    inertia.append(model.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(k_range, inertia, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia (error)")
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

**How to read it:** the line goes down fast at first, then slows down. The place where it starts to flatten is the elbow.
Here the bend is **gentle** (around K = 3 to 5) — real-life data is rarely perfect!

### Tool 2 — Silhouette Score
Gives a score from **-1 to +1**. Higher = groups are better separated. (We use 10,000 random rows to run fast.)

```python
sample = X_scaled[np.random.RandomState(42).choice(len(X_scaled), 10000, replace=False)]

scores = []
k_list = range(2, 9)

for k in k_list:
    model = KMeans(n_clusters=k, n_init=10, random_state=42).fit(X_scaled)
    labels = model.predict(sample)
    scores.append(silhouette_score(sample, labels))

plt.figure(figsize=(7, 4))
plt.plot(k_list, scores, marker="o", color="green")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Silhouette score")
plt.title("Silhouette Score")
plt.grid(True)
plt.show()

print(dict(zip(k_list, np.round(scores, 3))))
```

###  Which K do we choose?

- Scores are low everywhere (about 0.18–0.22) → the land types **overlap a lot**, there are no sharp separate islands.
- K = 2 has the best score, but only 2 groups is too simple to be useful.
- Elbow bends around 4.

 We choose **K = 4** because it gives a useful and easy-to-explain grouping.

>  **Teaching point:** In unsupervised learning there is **no single correct K**. You choose using the tools **and** common sense.

## Step 8️ — Train the final K-Means model

```python
K = 4
kmeans = KMeans(n_clusters=K, n_init=10, random_state=42)
kmeans.fit(X_scaled)

print("Model trained ")
```

### Add the cluster number to every row

```python
df["Cluster"] = kmeans.labels_
df[feature_cols + ["Cluster"]].head()
```

### How many rows are in each cluster?

```python
print(df["Cluster"].value_counts().sort_index())
```

```python
df["Cluster"].value_counts().sort_index().plot(kind="bar", color="orange", figsize=(6, 4))
plt.xlabel("Cluster")
plt.ylabel("Number of rows")
plt.title("Size of each cluster")
plt.xticks(rotation=0)
plt.show()
```

## Step 9️ — Understand each cluster (Profiling)

The computer only gives numbers 0, 1, 2, 3. **We** must find out what each group means.
We do this by looking at the **average** of every column inside each cluster.

```python
profile = df.groupby("Cluster")[feature_cols].mean().round(0)
profile
```

Let us compare the clusters on the most important columns as a graph.

```python
profile[["Elevation", "Horizontal_Distance_To_Roadways", "Horizontal_Distance_To_Fire_Points"]].plot(
    kind="bar", figsize=(9, 4))
plt.title("Average values per cluster")
plt.ylabel("Metres")
plt.xticks(rotation=0)
plt.show()
```

### Give simple names to clusters

Look at the table above and write a name for each cluster **in your own words**.
(Example: high elevation + far from roads = "High & remote land".)
From the table: Cluster 0 faces west (Aspect ~296); Cluster 1 faces east with the gentlest slope; Cluster 2 has the highest elevation and is far from water; Cluster 3 has the steepest slope, lowest elevation and is closest to roads. Edit the names below if you see it differently.

```python
# Names written after reading the 'profile' table above (you can change them)
cluster_names = {
    0: "West-facing, moderate land",
    1: "East-facing, gentle & flat land (biggest group)",
    2: "High & far from water",
    3: "Steep, low land near roads",
}
print(cluster_names)
```

## Step 10— Draw the clusters using PCA

We have 10 columns, but a graph can show only 2.
**PCA** squeezes 10 columns into 2 new columns while keeping as much information as possible.

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Information kept by 2 columns:", round(pca.explained_variance_ratio_.sum() * 100, 1), "%")
```

```python
# Plot 5000 random points (100000 would be too slow and messy)
idx = np.random.RandomState(1).choice(len(X_pca), 5000, replace=False)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_pca[idx, 0], X_pca[idx, 1],
                      c=kmeans.labels_[idx], cmap="viridis", s=8, alpha=0.7)
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Clusters drawn in 2D")
plt.legend(*scatter.legend_elements(), title="Cluster")
plt.show()
```

 You can see the colours form **zones**, but they touch each other. This matches the low silhouette score: the groups are *real but overlapping*.

## Step 11 — Predict on NEW (unseen) data

Now the most useful part. A new piece of land arrives that the model has **never seen**.
Which cluster does it belong to?

**Only 3 easy steps:**
1. Put the new data in a table with the **same 10 columns**
2. **Scale** it with the **same scaler** (use `transform`, never `fit` again)
3. Ask the model to **predict** the cluster

###  Step A — Add new data (3 new pieces of land)
You can type your own numbers here and change them.

```python
new_data = pd.DataFrame({
    "Elevation":                          [3300, 2100, 2800],
    "Aspect":                             [150,  45,   270],
    "Slope":                              [10,   30,   15],
    "Horizontal_Distance_To_Hydrology":   [200,  50,   400],
    "Vertical_Distance_To_Hydrology":     [20,   5,    80],
    "Horizontal_Distance_To_Roadways":    [3500, 800,  2000],
    "Hillshade_9am":                      [220,  200,  230],
    "Hillshade_Noon":                     [230,  215,  235],
    "Hillshade_3pm":                      [140,  120,  150],
    "Horizontal_Distance_To_Fire_Points": [2500, 600,  1800],
})
new_data
```

###  Step B — Scale the new data using the SAME scaler
 Use `transform` only. Never `fit_transform` on new data.

```python
new_scaled = scaler.transform(new_data[feature_cols])
print("Scaled ")
```

### Step C — Predict the cluster

```python
new_data["Predicted_Cluster"] = kmeans.predict(new_scaled)
new_data["Cluster_Name"] = new_data["Predicted_Cluster"].map(cluster_names)
new_data[["Elevation", "Slope", "Horizontal_Distance_To_Roadways", "Predicted_Cluster", "Cluster_Name"]]
```

### See where the new points fall on the picture
The big red stars are our **new data**. They land inside the cluster they were predicted for.

```python
new_pca = pca.transform(new_scaled)

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[idx, 0], X_pca[idx, 1], c=kmeans.labels_[idx], cmap="viridis", s=8, alpha=0.4)
plt.scatter(new_pca[:, 0], new_pca[:, 1], c="red", marker="*", s=400, edgecolors="black", label="New data")
for i, (x, y) in enumerate(new_pca):
    plt.text(x + 0.1, y + 0.1, f"New {i+1}", fontsize=11, fontweight="bold")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("New data on the cluster map")
plt.legend()
plt.show()
```

###  Bonus — One function to do everything
Wrap the 3 steps in one function. Now you can predict any new data in **one line**.

```python
def predict_cluster(new_df):
    scaled = scaler.transform(new_df[feature_cols])
    result = new_df.copy()
    result["Predicted_Cluster"] = kmeans.predict(scaled)
    result["Cluster_Name"] = result["Predicted_Cluster"].map(cluster_names)
    return result

# Test with just ONE new row
one_row = pd.DataFrame([[3000, 180, 12, 250, 40, 2400, 215, 225, 145, 2000]], columns=feature_cols)
predict_cluster(one_row)
```

## Summary

| Step | What we did |
|---|---|
| 1 | Loaded 1 lakh rows (no labels) |
| 2 | Chose 10 numeric columns |
| 3 | **Scaled** them with StandardScaler |
| 4 | Used **Elbow + Silhouette** to choose K |
| 5 | Trained **K-Means** (K = 4) |
| 6 | **Profiled** clusters (averages) |
| 7 | Drew them with **PCA** |
| 8 | **Predicted** the cluster of new unseen data |

###  Remember
- Unsupervised = **no answers given**, find patterns yourself
- **Always scale** before K-Means
- For new data: `scaler.transform` → `kmeans.predict`
- There is no perfect K, use tools + common sense
