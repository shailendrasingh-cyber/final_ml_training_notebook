## Part 1: Load & Inspect the Data (How to Spot Problems)
Before changing anything, we always **look** at the data first. Think of this like a doctor doing a check-up before giving medicine.


```python
import pandas as pd

# Load the CSV file into a DataFrame (a table in Python)
data = pd.read_csv('quikr_car.csv')

# See the first 5 rows
data.head()
```

```python
# Check how many rows and columns we have: (rows, columns)
data.shape
```

```python
# Check column names, missing (null) values, and data types
# Notice: year, Price, and kms_driven are 'object' (text), NOT numbers!
data.info()
```

### How do we know WHAT to change in each column?
We use `.unique()` on each column to see all distinct values and spot strange text or formatting.


```python
# Check unique values in 'year'
# Problem found: Contains text like '...', '150k', 'TOUR', 'sale', 'Zest' instead of years!
data['year'].unique()
```

```python
# Check unique values in 'Price'
# Problem found: Contains 'Ask For Price' and commas like '80,000'
data['Price'].unique()
```

```python
# Check unique values in 'kms_driven'
# Problem found: Contains ' kms' at the end, commas like '45,000 kms', 'nan', and 'Petrol'
data['kms_driven'].unique()
```

```python
# Check unique values in 'fuel_type'
# Problem found: Contains 'nan' (missing value), rest ('Petrol', 'Diesel', 'LPG') are fine!
data['fuel_type'].unique()
```

### Summary of Problems Found (Our Cleaning Checklist)
1. **`year`**: Has non-year text (`'...'`, `'TOUR'`, etc.) and is stored as text (`object`) -> Keep only numbers and convert to `int`.
2. **`Price`**: Has `'Ask For Price'` and commas (`','`) -> Remove `'Ask For Price'`, remove commas, and convert to `int`.
3. **`kms_driven`**: Has `' kms'`, commas (`','`), missing values, and `'Petrol'` -> Split at space to drop `'kms'`, remove commas, keep only numbers, and convert to `int`.
4. **`fuel_type`**: Has 1 missing (`NaN`) row left after earlier steps -> Drop missing rows.
5. **`name`**: Very long names -> Keep only the first 3 words for neatness.
6. **Outliers**: Check `describe()` for unrealistic prices (e.g., `85,00,003`) -> Remove extreme typo rows.

**What we DO NOT change:** `company` names and valid `fuel_type` categories (`Petrol`, `Diesel`, `LPG`) are already clean!


## Part 2: Step-by-Step Data Cleaning
First, always make a **backup copy** so we never lose the original data if we make a mistake.


```python
# Step 0: Make a safe backup copy
backup = data.copy()
```

### Step 1: Clean the `year` Column
- `data['year'].str.isnumeric()` checks every row: `True` if it is all digits (like `'2007'`), `False` if it has letters/symbols (like `'TOUR'`).
- `.astype(int)` converts text `'2007'` into a real number `2007`.


```python
# Works even if you accidentally run this cell twice!
data = data[data['year'].astype(str).str.isnumeric()]
data['year'] = data['year'].astype(int)

data.info()
```

### Step 2: Clean the `Price` Column
1. Remove rows where `Price == 'Ask For Price'` (we cannot do math on text).
2. Remove commas `','` using `.str.replace(',', '')` (because `'80,000'` cannot become an integer until the comma is gone).
3. Convert to `int` using `.astype(int)`.


```python
# Remove 'Ask For Price' rows
data = data[data['Price'] != 'Ask For Price']

# Remove commas and convert to integer
data['Price'] = data['Price'].str.replace(',', '').astype(int)

data['Price'].head()
```

### Step 3: Clean the `kms_driven` Column (Breaking Down the Chain!)
Look at a value like `'45,000 kms'`. We clean it in 4 tiny steps:
1. `.str.split(' ')` -> splits `'45,000 kms'` into a list `['45,000', 'kms']`.
2. `.str.get(0)` -> grabs the first item `'45,000'` (drops `'kms'`).
3. `.str.replace(',', '')` -> removes the comma to make `'45000'`.
4. `data['kms_driven'].str.isnumeric()` -> removes rows like `'Petrol'` or `NaN`, then `.astype(int)` turns it into a number.


```python
# 1. Remove missing (NaN) values in kms_driven first
data = data.dropna(subset=['kms_driven'])

# 2. Split by space, take the number part, and remove commas
data['kms_driven'] = data['kms_driven'].str.split(' ').str.get(0).str.replace(',', '')

# 3. Keep only numeric values (drops 'Petrol')
data = data[data['kms_driven'].str.isnumeric()]

# 4. Convert to integer
data['kms_driven'] = data['kms_driven'].astype(int)

data.info()
```

### Step 4: Clean the `fuel_type` Column
- `data['fuel_type'].isna()` finds missing values (`True` if missing).
- The `~` (tilde) means **NOT**. So `~data['fuel_type'].isna()` means: keep rows where `fuel_type` is **NOT** missing.


```python
# Remove rows where fuel_type is missing (NaN)
data = data[~data['fuel_type'].isna()]
```

### Step 5: Shorten Car `name` & Reset Row Numbers
- Keep only the first 3 words of each car name: split by space -> slice `[0:3]` -> join back with `' '`.
- `reset_index(drop=True)` fixes the row numbers (`0, 1, 2, ...`) after deleting rows.


```python
# Keep first 3 words of the car name
data['name'] = data['name'].str.split(' ').str.slice(0, 3).str.join(' ')

# Reset index numbers cleanly
data = data.reset_index(drop=True)

data.head()
```

### Step 6: Check for Outliers (Unrealistic Numbers)
Now that `year`, `Price`, and `kms_driven` are numbers, `data.describe()` shows us `min`, `max`, and `mean`. Look at `max` Price: `8.500003e+06` (over 85 Lakhs for a 2014 Mahindra XUV500!) — that is clearly a typo/outlier.


```python
# Summary statistics of numeric columns
data.describe()
```

```python
# 1. Remove 'Ask For Price' rows
data = data[data['Price'] != 'Ask For Price']

# 2. Remove commas and convert Price from text to integer
data['Price'] = data['Price'].str.replace(',', '').astype(int)

# 3. Now filter out the extreme outlier (60 Lakhs or more)
data = data[data['Price'] < 6e6].reset_index(drop=True)

# 4. Check the final clean dataset
data.shape
```

```python
# Save the cleaned dataset to a new CSV file
data.to_csv('cleaned_car.csv', index=False)
```

## Part 3: Finding Insights Using Only Pandas (No Graphs!)
Now that our data is clean, we can answer real business questions using simple Pandas methods.


```python
# Insight 1: Overall summary (average price, average kms, oldest & newest car)
data.describe().round(2)
```

```python
# Insight 2: Which car companies are listed most often? (Top 5)
data['company'].value_counts().head(5)
```

```python
# Insight 3: Petrol vs Diesel vs LPG — how many cars of each?
data['fuel_type'].value_counts()
```

```python
# Insight 4: Which fuel type has a higher average resale price?
data.groupby('fuel_type')['Price'].mean().round(0)
```

```python
# Insight 5: Top 5 most expensive and Top 5 cheapest cars
print('--- Top 5 Most Expensive Cars ---')
print(data.sort_values('Price', ascending=False)[['name', 'company', 'year', 'Price', 'kms_driven']].head(5))

print('\n--- Top 5 Cheapest Cars ---')
print(data.sort_values('Price', ascending=True)[['name', 'company', 'year', 'Price', 'kms_driven']].head(5))
```

```python
# Insight 6: Which car companies have the highest average price? (Top 5)
data.groupby('company')['Price'].mean().sort_values(ascending=False).round(0).head(5)
```

```python
# Insight 7: Which manufacturing years have the most cars for sale? (Top 5)
data['year'].value_counts().head(5)
```
