## Step 5 – Decide How to Handle Missing Values

There is no single rule: the decision depends on the column.

For this dataset:
- `email` → keep missing if unavailable
- `phone` → keep missing if unavailable
- `gender`, `city`, `department` → fill with `"Unknown"`
- `age`, `marks`, `attendance` → use a sensible numeric strategy
- `annual_salary` → median is safer than mean when outliers exist
- `admission_date` → keep missing if it cannot be recovered

```python
# Convert numeric columns to numeric data type
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["marks"] = pd.to_numeric(df["marks"], errors="coerce")
df["attendance"] = pd.to_numeric(df["attendance"], errors="coerce")
df["annual_salary"] = pd.to_numeric(df["annual_salary"], errors="coerce")
```

## Step 6 – Remove Extra Spaces and Standardize Text

A common real-world problem is that:

`Delhi`, ` Delhi`, `DELHI`, `delhi `

may represent the same category.

### Student Task
Clean the string columns using:
- `.str.strip()`
- `.str.lower()`

Then standardize known category variations.

```python
text_columns = ["student_name", "email", "gender", "city", "department", "course"]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

# Make selected categories lowercase for easier standardization
for col in ["gender", "city", "department", "course"]:
    df[col] = df[col].str.lower()

df[["student_name", "gender", "city", "department", "course"]].head(10)
```

```python
# Standardize category names
df["city"] = df["city"].replace({
    "bangalore": "bengaluru",
    "gurgaon": "gurugram"
})

df["department"] = df["department"].replace({
    "computer science": "cse",
    "information technology": "it",
    "electronics": "ece",
    "data science": "data science",
    "mechanical": "me"
})

df["course"] = df["course"].str.replace(" ", "", regex=False).str.lower()
df["course"] = df["course"].replace({
    "b.tech": "btech",
    "btech": "btech"
})

df[["city", "department", "course"]].value_counts().head(20)
```

## Step 8 – Clean Numeric Columns

The columns `age`, `marks`, `annual_salary`, and `attendance` may contain invalid values.

### Business rules
- Age should be between **17 and 35**
- Marks should be between **0 and 100**
- Attendance should be between **0 and 100**
- Salary should be greater than **0**

### Student Task
Identify invalid values first. Do not blindly delete rows.

```python
numeric_cols = ["age", "marks", "annual_salary", "attendance"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_cols].describe())
```

```python
# Trainer solution: convert impossible values to NaN
df.loc[~df["age"].between(17, 35), "age"] = np.nan
df.loc[~df["marks"].between(0, 100), "marks"] = np.nan
df.loc[df["annual_salary"] <= 0, "annual_salary"] = np.nan
df.loc[~df["attendance"].between(0, 100), "attendance"] = np.nan

# Refill values after invalid records were converted to NaN
df["age"] = df["age"].fillna(df["age"].median())
df["marks"] = df["marks"].fillna(df["marks"].median())
df["annual_salary"] = df["annual_salary"].fillna(df["annual_salary"].median())
df["attendance"] = df["attendance"].fillna(df["attendance"].median())
```

## Step 9 – Clean Email and Phone Numbers

### Email
A basic validation rule for this class:
- must contain `@`
- must contain `.`
- no spaces

### Phone
For this exercise, Indian student phone numbers should contain exactly **10 digits** and start with `6`, `7`, `8`, or `9`.

Do not build a perfect production validator yet. The goal is to understand cleaning logic.

```python
# Email cleaning
df["email"] = df["email"].str.strip().str.lower()

invalid_email = ~df["email"].str.match(r"^[^\s@]+@[^\s@]+\.[^\s@]+$", na=False)
print("Invalid emails:", invalid_email.sum())

df.loc[invalid_email, "email"] = np.nan

# Phone cleaning: remove spaces, hyphens and other non-digit characters
df["phone"] = df["phone"].astype("string").str.replace(r"\D", "", regex=True)

invalid_phone = ~df["phone"].str.match(r"^[6-9]\d{9}$", na=False)
print("Invalid phones:", invalid_phone.sum())

df.loc[invalid_phone, "phone"] = np.nan
```

# Part 2: Exploratory Data Analysis (EDA) & Visualization Lecture
### Using Matplotlib & Seaborn on the Cleaned Student Dataset

**Why do we visualize data?**
Summary tables like `.describe()` compress 50,000 rows into 8 numbers—which can hide skewed distributions, extreme outliers, and subgroup patterns. Visualization turns raw rows into immediate visual insight.

### The Two Libraries We Will Use
| Library | Role | Best For |
| :--- | :--- | :--- |
| **Matplotlib (`plt`)** | The foundational plotting engine in Python | Figure sizing (`figsize`), titles, axis labels, legends, annotations, and multi-plot grids (`subplots`) |
| **Seaborn (`sns`)** | Statistical visualization built on top of Matplotlib | Clean aesthetics, automatic aggregation, hue grouping, distributions (`histplot`, `boxplot`), and heatmaps |

> **Golden Rule of Plotting:** Always use `plt` (or `ax`) to control the **canvas, labels, and layout**, and use `sns` to draw **statistical data geometries** cleanly.

## Lesson 1 – Setup, Final Polish & Feature Engineering for Analysis

Before plotting, let's:
1. Import `matplotlib.pyplot` and `seaborn` and set a consistent, readable classroom theme.
2. Finish two tiny category cleanups (`m` → `male`, `f` → `female`) so our charts don't split the same gender into duplicate bars.
3. Engineer analytical features (`admission_year`, `admission_month`, `performance_category`, `attendance_category`) so we can slice and compare cohorts.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set global visual style for crisp, readable lecture charts
sns.set_theme(style="whitegrid", context="notebook", palette="Set2")
plt.rcParams["figure.dpi"] = 110

# Create an analysis copy and finalize gender standardization
eda_df = clean_df.copy()
eda_df["gender"] = eda_df["gender"].replace({"m": "male", "f": "female"})
eda_df["age"] = eda_df["age"].astype(int)

# Feature Engineering: Date parts
eda_df["admission_year"] = eda_df["admission_date"].dt.year
eda_df["admission_month"] = eda_df["admission_date"].dt.month_name().str[:3]

# Feature Engineering: Performance & Attendance Tiers using pd.cut
eda_df["performance_category"] = pd.cut(
    eda_df["marks"],
    bins=[0, 40, 60, 75, 100],
    labels=["Needs Improvement (<40)", "Average (40-60)", "Good (60-75)", "Excellent (75+)"],
    include_lowest=True
)

eda_df["attendance_category"] = pd.cut(
    eda_df["attendance"],
    bins=[0, 75, 85, 100],
    labels=["Low (<75%)", "Moderate (75-85%)", "High (85%+)"],
    include_lowest=True
)

eda_df[["student_id", "gender", "marks", "performance_category", "attendance", "attendance_category", "admission_year"]].head()
```

## Lesson 2 – Univariate Analysis (Categorical Data): Count Plots & Bar Charts

**Concept:** *Univariate* means analyzing **one variable at a time**. For categorical variables (`department`, `course`, `performance_category`), we want to see **frequency counts**.

**Key Techniques Taught Here:**
- `plt.subplots(1, 2, figsize=(14, 5))`: Placing two related charts side-by-side.
- `sns.countplot()`: Automatically counts occurrences of each category.
- `order=...`: Sorting bars from highest to lowest frequency (a best practice for readability).
- `ax.bar_label()`: Adding exact numbers on top of bars so readers don't have to guess.

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Student Count by Department (sorted descending)
dept_order = eda_df["department"].value_counts().index
sns.countplot(data=eda_df, x="department", order=dept_order, ax=axes[0], palette="viridis")
axes[0].set_title("Student Enrollment by Department", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Department")
axes[0].set_ylabel("Number of Students")
for container in axes[0].containers:
    axes[0].bar_label(container, fmt="%d", padding=3, fontsize=9)

# Plot 2: Student Count by Performance Category
sns.countplot(data=eda_df, x="performance_category", ax=axes[1], palette="coolwarm")
axes[1].set_title("Distribution of Academic Performance Tiers", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Performance Tier")
axes[1].set_ylabel("Number of Students")
axes[1].tick_params(axis="x", rotation=15)
for container in axes[1].containers:
    axes[1].bar_label(container, fmt="%d", padding=3, fontsize=9)

plt.tight_layout()
plt.show()
```

## Lesson 3 – Univariate Analysis (Numeric Data): Histograms, KDE & Mean vs. Median

**Concept:** For continuous numeric variables (`marks`, `attendance`), we examine the **shape of the distribution** (bell-shaped/normal, skewed, or bimodal).

**Key Techniques Taught Here:**
- `sns.histplot(..., kde=True)`: Combines binned frequency bars with a smooth **Kernel Density Estimate (KDE)** curve.
- `ax.axvline()`: Drawing vertical reference lines for **Mean** vs. **Median**.
- **Teaching Insight:** When Mean ≈ Median, the data is symmetric. When Mean differs sharply from Median, outliers or skewness are pulling the average!

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Marks Distribution
sns.histplot(data=eda_df, x="marks", bins=30, kde=True, color="steelblue", ax=axes[0])
axes[0].axvline(eda_df["marks"].mean(), color="crimson", linestyle="--", linewidth=2, label=f"Mean: {eda_df['marks'].mean():.1f}%")
axes[0].axvline(eda_df["marks"].median(), color="darkgreen", linestyle="-", linewidth=2, label=f"Median: {eda_df['marks'].median():.1f}%")
axes[0].set_title("Distribution of Student Marks (%)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Marks (%)")
axes[0].legend()

# Attendance Distribution
sns.histplot(data=eda_df, x="attendance", bins=30, kde=True, color="teal", ax=axes[1])
axes[1].axvline(eda_df["attendance"].mean(), color="crimson", linestyle="--", linewidth=2, label=f"Mean: {eda_df['attendance'].mean():.1f}%")
axes[1].axvline(eda_df["attendance"].median(), color="orange", linestyle="-", linewidth=2, label=f"Median: {eda_df['attendance'].median():.1f}%")
axes[1].set_title("Distribution of Student Attendance (%)", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Attendance (%)")
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Lesson 4 – Detecting & Visualizing Outliers: Boxplots on `annual_salary`

**Concept:** Remember in Step 8 when `annual_salary` had a mean of ~₹69,596 but a median of ~₹40,109, and a max of ₹9,999,999? A histogram of raw salary would look squashed against the left wall because of extreme outliers!

**Key Techniques Taught Here:**
- `sns.boxplot()`: Visualizes the **Interquartile Range (IQR)** — 25th percentile ($Q_1$), Median ($Q_2$), 75th percentile ($Q_3$), and outlier dots beyond $1.5 \times \text{IQR}$.
- **Before vs. After Outlier Capping (99th Percentile):** Comparing the raw salary boxplot alongside a trimmed/capped view (`<= 99th percentile`) shows students how analysts inspect skewed financial data.

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Left: Raw Annual Salary (shows extreme placeholder outliers like 9,999,999)
sns.boxplot(data=eda_df, x="annual_salary", color="salmon", ax=axes[0])
axes[0].set_title("Raw Annual Salary (Squashed by Extreme Outliers)", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Annual Salary (INR)")

# Right: Salary filtered below 99th percentile to reveal the true distribution
salary_99 = eda_df["annual_salary"].quantile(0.99)
filtered_salary = eda_df[eda_df["annual_salary"] <= salary_99]

sns.boxplot(data=filtered_salary, x="annual_salary", color="mediumseagreen", ax=axes[1])
axes[1].set_title(f"Annual Salary (<= 99th Percentile: ₹{salary_99:,.0f})", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Annual Salary (INR)")

plt.tight_layout()
plt.show()
```

## Lesson 5 – Bivariate Analysis (Categorical vs. Numeric): Boxplots & Violin Plots

**Concept:** *Bivariate* means comparing **two variables together**. Here we ask: *How do Marks and Salary vary across Courses and Departments?*

**Key Techniques Taught Here:**
- `sns.boxplot(x=category, y=numeric)`: Compares medians and spread across groups.
- `sns.violinplot(x=category, y=numeric)`: Combines a boxplot with a KDE density shape on both sides so we see where students cluster.
- `hue=...`: Adding a third dimension (like `course`) inside the same chart.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Marks distribution by Course (Violin Plot)
sns.violinplot(data=eda_df, x="course", y="marks", inner="quartile", palette="Pastel1", ax=axes[0])
axes[0].set_title("Marks Distribution by Course (Violin + Quartiles)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Course")
axes[0].set_ylabel("Marks (%)")

# Plot 2: Salary (<=99th pct) across Departments split by Course
sns.boxplot(
    data=filtered_salary[filtered_salary["department"] != "unknown"],
    x="department",
    y="annual_salary",
    hue="course",
    palette="Set2",
    ax=axes[1]
)
axes[1].set_title("Annual Salary by Department & Course", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Department")
axes[1].set_ylabel("Annual Salary (INR)")
axes[1].legend(title="Course", loc="upper right")

plt.tight_layout()
plt.show()
```

## Lesson 6 – Bivariate Analysis (Numeric vs. Numeric): Scatter & Regression Plots

**Concept:** Does higher attendance actually lead to higher marks? When both variables are numeric, we use a **Scatter Plot**.

**Why Sampling Matters with 50,000 Rows:**
Plotting 50,000 dots on top of each other causes *overplotting* (a solid blob of ink) and slows down rendering. Taking a random sample (`eda_df.sample(1500, random_state=42)`) reveals the true pattern clearly and quickly.

**Key Techniques Taught Here:**
- `sns.regplot()`: Draws scatter points plus a linear **regression trendline** and 95% confidence band.
- `scatter_kws={"alpha": 0.35}`: Makes points semi-transparent so overlapping clusters stand out.

```python
sample_df = eda_df.sample(n=1500, random_state=42)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Attendance vs Marks with Regression Trendline
sns.regplot(
    data=sample_df,
    x="attendance",
    y="marks",
    scatter_kws={"alpha": 0.35, "color": "royalblue"},
    line_kws={"color": "crimson", "linewidth": 2},
    ax=axes[0]
)
axes[0].set_title("Attendance % vs. Marks % (Sample n=1,500)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Attendance (%)")
axes[0].set_ylabel("Marks (%)")

# Plot 2: Marks vs Annual Salary colored by Course
sample_sal = filtered_salary.sample(n=1500, random_state=42)
sns.scatterplot(
    data=sample_sal,
    x="marks",
    y="annual_salary",
    hue="course",
    alpha=0.55,
    palette="Dark2",
    ax=axes[1]
)
axes[1].set_title("Marks % vs. Annual Salary by Course", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Marks (%)")
axes[1].set_ylabel("Annual Salary (INR)")

plt.tight_layout()
plt.show()
```

## Lesson 7 – Multivariate Analysis: Correlation Heatmap & Cross-Tab Heatmap

**Concept:** Heatmaps turn matrices of numbers into color gradients so patterns jump out instantly.

**Key Techniques Taught Here:**
1. **Correlation Matrix (`df.corr()`)**: Measures linear relationships between numeric columns from `-1` (strong negative) to `+1` (strong positive), with `0` meaning no linear relationship.
2. **Pivot / Cross-Tab Heatmap (`pd.crosstab(..., normalize='index')`)**: Shows the percentage breakdown of performance tiers inside each department.
- `annot=True`: Prints the exact numbers inside every cell.
- `cmap="coolwarm"` / `cmap="YlGnBu"`: Diverging vs. sequential color palettes.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Numeric Correlation Heatmap (using filtered_salary so extreme 9.99M outliers don't distort correlation)
corr_cols = ["age", "marks", "attendance", "annual_salary"]
corr_matrix = filtered_salary[corr_cols].corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=1,
    ax=axes[0]
)
axes[0].set_title("Correlation Matrix of Numeric Features", fontsize=13, fontweight="bold")

# Plot 2: Department vs Performance Tier (% Row-Normalized Crosstab)
dept_perf_pct = pd.crosstab(
    eda_df["department"],
    eda_df["performance_category"],
    normalize="index"
) * 100

sns.heatmap(
    dept_perf_pct,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu",
    linewidths=1,
    ax=axes[1]
)
axes[1].set_title("Performance Tier Share (%) Within Each Department", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Performance Category")
axes[1].set_ylabel("Department")

plt.tight_layout()
plt.show()
```

## Lesson 8 – Time-Series & Geographic Trends: Line Charts & Horizontal Bar Charts

**Concept:** Different questions require different orientations:
- **Time trends** (`admission_year`, `admission_month`) should flow **left-to-right** using a **Line Chart** (`sns.lineplot`).
- **Long category names** (`city` with 13+ Indian cities) should be plotted as **Horizontal Bar Charts** (`y="city"`) so city names remain horizontal and easy to read without neck strain!

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Monthly Admission Trend Across Years (Line Plot)
month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
monthly_admissions = (
    eda_df.dropna(subset=["admission_year"])
    .groupby(["admission_year", "admission_month"])
    .size()
    .reset_index(name="student_count")
)
monthly_admissions["admission_year"] = monthly_admissions["admission_year"].astype(int)
monthly_admissions["admission_month"] = pd.Categorical(monthly_admissions["admission_month"], categories=month_order, ordered=True)
monthly_admissions = monthly_admissions.sort_values(["admission_year", "admission_month"])

sns.lineplot(
    data=monthly_admissions,
    x="admission_month",
    y="student_count",
    hue="admission_year",
    marker="o",
    palette="tab10",
    ax=axes[0]
)
axes[0].set_title("Monthly Student Admissions by Year", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Admissions Count")
axes[0].legend(title="Year")

# Plot 2: Top 10 Cities by Student Count (Horizontal Bar Chart)
top_cities = eda_df[eda_df["city"] != "unknown"]["city"].value_counts().head(10)
sns.barplot(x=top_cities.values, y=top_cities.index.str.title(), palette="mako", ax=axes[1])
axes[1].set_title("Top 10 Cities by Student Enrollment", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Number of Students")
axes[1].set_ylabel("City")
for container in axes[1].containers:
    axes[1].bar_label(container, fmt="%d", padding=4, fontsize=9)

plt.tight_layout()
plt.show()
```

## Lesson 9 – Putting It All Together: An Executive Summary Dashboard (`GridSpec`)

**Key Techniques Taught Here:**

```python
fig = plt.figure(figsize=(16, 11))
gs = fig.add_gridspec(2, 2, hspace=0.35, wspace=0.25)

# Panel 1 (Top-Left): Gender Breakdown (Pie / Donut Chart in Matplotlib)
ax1 = fig.add_subplot(gs[0, 0])
gender_counts = eda_df["gender"].value_counts()
wedges, texts, autotexts = ax1.pie(
    gender_counts.values,
    labels=gender_counts.index.str.title(),
    autopct="%1.1f%%",
    startangle=140,
    colors=sns.color_palette("Set2"),
    wedgeprops={"width": 0.45, "edgecolor": "white"}
)
ax1.set_title("1. Student Gender Composition", fontsize=12, fontweight="bold")

# Panel 2 (Top-Right): Average Marks by Attendance Category
ax2 = fig.add_subplot(gs[0, 1])
sns.barplot(
    data=eda_df,
    x="attendance_category",
    y="marks",
    estimator="mean",
    errorbar="sd",
    palette="Blues_d",
    ax=ax2
)
ax2.set_title("2. Average Marks (±1 SD) by Attendance Tier", fontsize=12, fontweight="bold")
ax2.set_xlabel("Attendance Tier")
ax2.set_ylabel("Mean Marks (%)")
ax2.set_ylim(0, 105)
for container in ax2.containers:
    ax2.bar_label(container, fmt="%.1f%%", label_type="center", color="white", fontweight="bold")

# Panel 3 (Bottom-Left): Median Salary by Department (Excluding 'unknown' & >99th pct outliers)
ax3 = fig.add_subplot(gs[1, 0])
dept_sal = (
    filtered_salary[filtered_salary["department"] != "unknown"]
    .groupby("department")["annual_salary"]
    .median()
    .sort_values(ascending=False)
)
sns.barplot(x=dept_sal.index.str.upper(), y=dept_sal.values, palette="Greens_r", ax=ax3)
ax3.set_title("3. Median Annual Salary by Department", fontsize=12, fontweight="bold")
ax3.set_xlabel("Department")
ax3.set_ylabel("Median Salary (INR)")
for container in ax3.containers:
    ax3.bar_label(container, fmt="₹%.0f", padding=3, fontsize=9)

# Panel 4 (Bottom-Right): Age Distribution by Course
ax4 = fig.add_subplot(gs[1, 1])
sns.boxplot(data=eda_df, x="course", y="age", palette="Set3", ax=ax4)
ax4.set_title("4. Student Age Spread by Course", fontsize=12, fontweight="bold")
ax4.set_xlabel("Course")
ax4.set_ylabel("Age (Years)")

fig.suptitle("B.Tech & Higher-Ed Student Cohort: Exploratory Analysis Dashboard", fontsize=16, fontweight="bold", y=0.98)
plt.savefig("student_eda_dashboard.png", dpi=200, bbox_inches="tight")
plt.show()
```
